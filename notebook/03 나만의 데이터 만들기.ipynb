```python
born_datetime = pd.to_datetime(scientists['Born'], format='%Y-%m-%d') # 파이썬 날짜 포맷 참조
born_datetime
```

```python
scientists['Born']
```

```python
died_datetime = pd.to_datetime(scientists['Died'], format='%Y-%m-%d')
died_datetime
```

```python
scientists['born-dt'], scientists['died-dt'] = (born_datetime, died_datetime)
```

```python
scientists
```

```python
scientists.dtypes
```

```python
scientists
```

```python
import numpy as np

scientists = scientists.assign(
    age_days_assign = scientists['died-dt'] - scientists['born-dt'],
    age_year_assign = lambda df_:
     (df_["age_days_assign"].dt.days/365).apply(np.floor),
    )

scientists
```

```python
scientists.dtypes
```

```python
# 무작위 섞기(sample), 값을 섞기
scientists['Age'] = scientists['Age'].sample(frac=1, random_state=42).values
scientists
```

```python
# index 섞기

scientists['Age'] = scientists['Age'].sample(frac=1, random_state=42)
scientists
```

```python
scientists_dropped = scientists.drop(['Age'], axis="columns")
scientists_dropped
```
