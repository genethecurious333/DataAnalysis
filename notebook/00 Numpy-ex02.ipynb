```python
ratings = np.array([
    4.5,
    3.8,
    5.0,
    2.5,
    4.7,
    3.2,
    4.9   
])
# 평점 4.0 이상만 추출
ratings[ratings >=4.0]
```

```python
# 평점 3.0 미만만 추출
ratings[ratings < 3.0]
```

```python
ratings[ratings >= 4.5]
```

```python
# 소숫점이하에 0을 표시하고 싶을때 
np.set_printoptions(formatter={'float' : '{:0.1f}'.format})
# 평점 4.0 이상만 추출
print(ratings[ratings >= 4.5])
                               
```
