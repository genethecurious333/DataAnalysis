```python
weather = pd.read_csv('../data/weather.csv')
print(weather.iloc[:5, :11])
```

```python
weather_melt = weather.melt(
    id_vars=["id", "year", "month", "element"], # id_vars는 melt하지 않는 부분 지정하는 코드임, 인덱스, 리스트[]로 묶어 준다
    var_name="day",
    value_name="temp"
)

print(weather_melt)
```

```python
weather_tidy = weather_melt.pivot_table(
    index = ['id', 'year', 'month', 'day'],
    columns= 'element',
    values='temp'
)
# melt된 요소를 다시 coloumn에 올리는 것이 pivot이며, elements를 column에 올리고, 그 elements의 값(values)는 temp가 들고 있다는 의미임
print(weather_tidy)
```

```python
weather_tidy_flat = weather_tidy.reset_index()  # 데이터를 평탄화한다, 생략된 데이터를 모두 채워주기
print(weather_tidy_flat)
```

```python
# melt된것을 다시 올리고(pivot), 평탄화까지 일괄 처리한다

weather_tidy = (
    weather_melt
    .pivot_table(
        index = ['id', 'year', 'month', 'day'],
        columns= 'element',
        values='temp')
    .reset_index()
)

print(weather_tidy)
```
