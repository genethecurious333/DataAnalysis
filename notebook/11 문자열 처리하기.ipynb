# 정규식 문자열 처리하기

```python
import re

tele_num = '1234567890'
```

```python
m = re.match(pattern= '\d\d\d\d\d\d\d\d\d\d', string=tele_num)

print(type(m))
```

```python
print(m)
```

```python
print(bool(m))
```

```python
if m:
    print('match')
else:
    print('no match')
```

```python
print(m.start())
print(m.end())
print(m.span())
print(m.group())
```

```python
tele_num_spaces = '123 456 7890'
```

```python
m = re.match(pattern='\d{10}', string=tele_num_spaces)

print(m)
```

```python
if m:
    print('match')
else:
    print('no match')
```

```python
p = r'\d{3}\s?\d{3}\s?\d{4}'
m = re.match(pattern=p, string=tele_num_spaces)
print(m)
```

```python
tele_num_space_paren_dash = '(123) 456-7890'
p = r'\(?\d{3}\)?\s?\d{3}\s?-?\d{4}'
m = re.match(pattern=p, string=tele_num_space_paren_dash)

print(m)
```

```python
cnty_tele_num_space_paren_dash = '+1 (123) 456-7890'
p = r'\+?1\s?\(?\d{3}\)?\s?\d{3}\s?-?\d{4}'
m = re.match(pattern=p, string=cnty_tele_num_space_paren_dash)

print(m)
```

```python
p = (
    r'\+?'  # +가 0개 또는 1개
    r'1'    # 숫자 1
    r'\s?'  # 공백 문자가 0개 또는 1개
    r'\(?'  # ( 문자가 0개 또는 1개
    r'\d{3}'# 숫자가 3개
    r'\)?'  # ) 문자가 0개 또는 1개
    r'\s?'  # 공백 문자가 0개 또는 1개
    r'\d{3}'# 숫자 3개
    r'\s?'  # 공백 문자가 0개 또는 1개
    r'-?'   # - 문자가 0개 또는 1개
    r'\d{4}' # 숫자가 4개
)

print(p)
```

```python
cnty_tele_num_space_paren_dash = '+1 (123) 456-7890'
m = re.match(pattern=p, string=cnty_tele_num_space_paren_dash)

print(m)
```

```python
s = (
    '14 Ncuti Gatwa,'
    '13 Jodie Whittaker, war John Hurt, 12 Peter Capaldi,'
    '11 Matt Smith, 10 David Tennant, 9 Christopher Eccleston'
)

print(s)
```

```python
p = '\d+' # 숫자만 뽑는다
m = re.findall(pattern=p, string=s)

print(m)
```

```python
multi_str = """Guard: What? Ridden on a horse?
King Arthur: Yes!
Guard: You're using coconuts!
King Arthur: What?
Guard: You've got ... coconut[s] and you're bangin' 'em together
"""

p = '\w+\s?\w+:\s?'
s = re.sub(pattern=p, string=multi_str, repl='')

print(s)
```

```python
guard = s.splitlines()[::2]
kinga = s.splitlines()[1::2]

print(guard)
```

```python
p = re.compile('\d{10}')
s = '1234567890'
m = p.match(s)

print(m)
```

```python
p = re.compile('\d+')
s = (
    '14 Ncuti Gatwa,'
        '13 Jodie Whittaker, war John Hurt, 12 Peter Capaldi,'
        '11 Matt Smith, 10 David Tennant, 9 Christopher Eccleston'
)

m = p.findall(s)

print(m)
```

```python
p = re.compile('\w+\s?\w+:\s?')
s = "Guard: You're using coconuts!"
m = p.sub(string=s, repl='')

print(m)
```
