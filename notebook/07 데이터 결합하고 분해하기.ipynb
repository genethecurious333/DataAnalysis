```python
# billboard 데이터에서 트랙의 주별 순위를 매기기 : 데이터 정규화
```

```python
import pandas as pd

billboard = pd.read_csv('../data/billboard.csv')

billboard_long = billboard.melt(
    id_vars = ['year', 'artist', 'track', 'time', 'date.entered'],
    var_name = 'week', 
    value_name ='rating'
)

print(billboard_long)
```

```python
print(billboard_long.loc[billboard_long.track == 'Loser'])
```

```python
billboard_songs = billboard_long [
    ['year', 'artist', 'track', 'time', 'date.entered']
]

print(billboard_songs.shape)
```

```python
billboard_songs
```

```python
# 중복 제거 - 정규과 Normalization 과정
billboard_songs = billboard_songs.drop_duplicates()
print(billboard_songs.shape)
```

```python
# column에 id를 추가한다, index에 + 1을 한다.

billboard_songs['id'] = billboard_songs.index + 1 
print(billboard_songs)
```

```python
billboard_ratings = billboard_songs.merge(
    billboard_long, on=['year', 'artist', 'track', 'time', 'date.entered']
)

print(billboard_ratings.shape)
```

```python
print(billboard_ratings)
```

```python
billboard_ratings = billboard_ratings[['id', 'week', 'rating']]

display(billboard_ratings)
```
