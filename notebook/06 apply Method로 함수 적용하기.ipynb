### apply() 메서드 사용하기

```python
import pandas as pd
df = pd.DataFrame({
    "a": [10, 20, 30],
    "b": [20, 30, 40]
})

print(df)
```

```python
def my_sq(x):
    return x**2

def avg_2(x, y):
    return (x + y) / 2
```

```python
sq_result = df['a'].apply(my_sq)  # my_sq는 callback 함수임
sq_result
```

```python
avg_result = df['a'].apply(avg_2, y=100)
avg_result
```

```python
avg_result = df['a'].apply(avg_2, args=(200,))  
# args는 tuple형태로 넘겨야한다, apply함수는 x와 y값이 필요하며, df['a']가 x, args=(200,)이 y값임
avg_result
```

```python
def my_exp(x, e):
    return x ** e

ex = df['a'].apply(my_exp, e=3)
print(ex)
```

```python
def avg_3_apply(col):
    x = col[0]
    y = col[1]
    z = col[2]
    return (x + y + z) / 3

print(df.apply(avg_3_apply))
```

```python
df.apply(avg_3_apply, axis=0)
```

```python
def avg_2_apply(row):
    x = row.iloc[0]
    y = row.iloc[1]
    return (x + y) / 2

print(df.apply(avg_2_apply, axis=1))
```

람다 함수 사용하기
# 람다는 1회용함수, 즉석 함수다
# lambda를 쓴다, 그다음 input을 주고, return을 준다.

```python
def my_sq(x):
    return x**2

df['a_sq'] = df['a'].apply(my_sq)
print(df)
```

```python
# 람다를 쓰고, 매개변수(parameter)를 쓰고 콜론 : 을 쓴후, 계산식을 넣는다.

df['a_sq_lamb'] = df['a'].apply(lambda x: x **2)
print(df)
```


벡터화하기


```python
def avg_2(x, y):
    return (x + y) / 2
```

```python
print(avg_2(df['a'], df['b']))
```

```python
import numpy as np

def avg_2_mod(x, y):
    if (x == 20):
        return(np.nan)
    else:
        return (x + y) / 2
```

```python
# print(avg_2_mod(df['a'], df['b'])) # 오류
```

```python
print(avg_2_mod(20, 30))
```

```python
import numpy as np

avg_2_mod_vec = np.vectorize(avg_2_mod)

print(avg_2_mod_vec(df['a'], df['b']))
```

```python
@np.vectorize
def v_avg_2_mod(x, y):
    if (x == 20):
        return(np.nan)
    else:
        return (x + y) / 2

print(v_avg_2_mod(df['a'], df['b']))
```

```python
# numba 로 벡터화하기

import numba

@numba.vectorize
def v_avg_2_numba(x, y):
    if (int(x) == 20):
        return(np.nan)
    else:
        return (x + y) / 2
```
