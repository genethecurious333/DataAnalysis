```python
ebola = pd.read_csv(
    '../data/country_timeseries.csv',
    parse_dates=['Date'],
    index_col='Date',
)

print(ebola.iloc[:, :4])
```

```python
new_idx = pd.date_range(ebola.index.min(), ebola.index.max()) # 제일 먼저 발병한 나라, 제일 마지막에 발병한 나라
print(new_idx)
```

```python
new_idx = reversed(new_idx)
print(new_idx)
```

```python
ebola = ebola.reindex(new_idx)
```

```python
print(ebola.iloc[:, :4])
```

```python
last_valid = ebola.apply(pd.Series.last_valid_index)
print(last_valid)
```

```python
earlist_date = ebola.index.min() # 전체 최초 발병일
print(earlist_date)
```

```python
shift_values = last_valid - earlist_date
print(shift_values)
```

```python
ebola_dict = {}

for idx, col in enumerate(ebola):
    d = shift_values.iloc[idx].days
    shifted = ebola[col].shift(d)
    ebola_dict[col] = shifted

print(ebola_dict)
```

```python
ebola_shift = pd.DataFrame(ebola_dict)
```

```python
print(ebola_shift.tail())
```

```python
ebola_shift.index = ebola_shift['Day']
ebola_shift = ebola_shift.drop(['Day'], axis='columns')
print(ebola_shift.tail())
```
