```python
# 결측값 개수 구하기
```

```python
ebola = pd.read_csv('../data/country_timeseries.csv')

print(ebola.count()) # nan 제외
```

```python
num_rows = ebola.shape[0]
num_missing = num_rows - ebola.count()
print(num_missing)
```

```python
import numpy as np

print(np.count_nonzero(ebola.isnull())) # 전체 DataFrame의 결측치 개수
```

```python
print(np.count_nonzero(ebola['Cases_Guinea'].isnull())) # 특정 국가의 결측치 개수
```

```python
cnts = ebola.Cases_Guinea.value_counts(dropna=False)
print(cnts)
```

```python
print(cnts.loc[pd.isnull(cnts.index)])
```

```python
print(ebola.Cases_Guinea.isnull().sum())
```

```python
cnts.index
```

```python
pd.isnull(cnts.index)
```

```python
cnts.loc[pd.isnull(cnts.index)]
```

```python
print(ebola.Cases_Guinea.isnull().sum())
```

```python
# 결측값 대체하기
```

```python
print(ebola.fillna(0).iloc[:, 0:5])
```

```python
# 정방향으로 채우기 forward fill
# print(ebola.fillna(method='ffill).iloc[:, 0:5])
print(ebola.ffill().iloc[:, 0:5])
```

```python
# print(ebola.interpolate().iloc[:, 0:5])
# 올바른 방법: 수치형(number) column만 선택 후 보간
print(ebola.iloc[:, 0:5].select_dtypes(include='number').interpolate())
```

```python
# Pandas 내장 na 결측값 살펴보기
```

```python
scientists = pd.DataFrame(
    {
        'Name' : ['Rosaline Franklin', 'William Gosset'],
        'Occupation': ['Chemist', 'Statistician'],
        'Born' : ['1920-07-25', '1876-06-13'],
        'Died' : ['1958-04-16', '1937-10-16'],
        'Age' : [37, 61],
    }
)

print(scientists)
```

```python
print(scientists.dtypes)
```

```python
scientists.loc[1, 'Name'] = pd.NA
scientists.loc[1, 'Age'] = pd.NA
print(scientists)
```

```python
print(scientists.dtypes)
```
