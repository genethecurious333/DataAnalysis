```python
# 년도별 기대수명 평균 산출하기
print(df.groupby('year')['lifeExp'].mean())
```

```python
# 년도별 인구의 평균

df.groupby('year')['pop'].mean()
```

```python
df.groupby('year')['gdpPercap'].mean()
# 연도별 gdp평균
```

```python
# 여러개열을 그룹화할때는 리스트[]로 묶어준다
df.groupby(['year', 'continent'])[['lifeExp', 'gdpPercap']].mean()
```

```python
# 계층구조를 없애고 데이터 프레임을 평탄화flat 하기
df.groupby(['year', 'continent'])[['lifeExp', 'gdpPercap']].mean().reset_index()
# Method Chain
```

```python
multi_group_var = (df
                   .groupby(['year', 'continent'])[['lifeExp', 'gdpPercap']]
                   .mean()
                   .reset_index()
                   )
multi_group_var
```

```python
# 그룹화한 데이터 갯수 구하기
# nunique() : 중복 제외한 갯수
df.groupby('continent')['country'].nunique()
```

```python
# value_counts() : 지정한 열이나 행의 갯수 (중복 포함 카운트)
df.groupby('continent')['country'].value_counts()
```

```python
# 년도별 평균 기대 수명을 구하기
global_yearly_life_expectancy = df.groupby('year')['lifeExp'].mean()
global_yearly_life_expectancy
```

```python
global_yearly_life_expectancy.plot()
# 간단한 그래프 그리스 : .plot()
```
