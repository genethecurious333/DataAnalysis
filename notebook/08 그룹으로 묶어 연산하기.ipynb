```python
def my_mean(values):
    n = len(values)        # 숫자 개수를 구합니다.
    sum = 0                # 합계를 0으로 초기화한다
    for value in values:   
        sum += value        # 각 값을 더한다
    return sum / n          # 합계를 숫자 개수로 나눈값을 반환한다
```

```python
agg_my_mean = df.groupby('year')['lifeExp'].agg(my_mean)
print(agg_my_mean)
```

```python
def my_mean_diff(values, diff_value):
    n = len(values)
    sum = 0 
    for value in values:
        sum += value
    mean = sum / n
    return(mean - diff_value)
```

```python
global_mean = df['lifeExp'].mean()
print(global_mean)
```

```python
agg_mean_diff = (
    df
    .groupby('year')
    ['lifeExp']
    .agg(my_mean_diff, diff_value=global_mean)
)

print(agg_mean_diff)
```

```python
# 여러 개의 집계 함수 한번에 사용하기

gdf = (
    df
    .groupby('year')
    ['lifeExp']
    # .agg([np.count_nonzero, np.mean, np.std])
    .agg(['count', 'mean', 'std'])
)

print(gdf)
```

```python
# agg()나 aggregate() Method에 Dictionary 사용하기
# 각 열마다 다른 집계함수를 적용할때
```

```python
gdf_dict = df.groupby('year').agg(
    {
        'lifeExp' : 'mean',
        'pop' : 'median',
        'gdpPercap' : 'median' # median 중간값
    }
)

print(gdf_dict)
```

```python
# Series에 사용하기

gdf = (
    df
    .groupby('year')
    ['lifeExp']
    .agg(
        [
            np.count_nonzero,
            np.mean,
            np.std,
        ]
    )
    .rename(
        columns={
            'count_nonzero' : 'count',
            'mean' : 'avg',
            'std' : 'std_dev',
        }
    )
    .reset_index()  # 평탄화한 DataFrame 반환하기
)

print(gdf)
```

```python
tips_10['fill_total_bill'] = total_bill_group_mean
print(tips_10[['sex', 'total_bill', 'fill_total_bill']])
```

```python
for sex_group in grouped:

    # 객체 자료형(Tuple)
    print(f'the type is: {type(sex_group)}\n')

    # 객체 길이 (2)
    print(f'the length is: {len(sex_group)}\n')

    # 첫 번째 요소
    first_element = sex_group[0]
    print(f'the first element is: {first_element}\n')

    # 첫 번째 요소의 자료형(문자열)
    print(f'it has a type of: {type(sex_group[0])}\n')

    # 두번째 요소
    second_element = sex_group[1]
    print(f'the second element is: \n{second_element}\n')

    # 두 번째 요소의 자료형 (DataFrame)
    print(f'it has a type of: {type(second_element)}\n')

    # 그룹 출력
    print(f'what we have:')
    print(sex_group)

    # for문 중단, 하나의 데이터만 꺼내 보겠다는 의미
    break 
```

```python
intv_df = pd.read_csv('../data/epi_sim.zip')
print(intv_df)
```

```python
count_only = (
    intv_df
    .groupby(['rep', 'intervened', 'tr'])
    ['ig_type']
    .count()
)

print(count_only)
```

```python
print(type(count_only))
```

```python
print(count_only.index)
```

```python
count_mean = count_only.groupby(level=[0, 1, 2]).mean()
print(count_mean.head())
```

```python
count_mean = (
    intv_df
    .groupby(['rep', 'intervened', 'tr'])['ig_type']
    .count()
    .groupby(level=[0, 1, 2])
    .mean()
)

print(count_mean.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

fig = sns.lmplot(
    data = count_mean.reset_index(),
    x='intervened',  # 개입한 시간
    y='ig_type',     # 개입 횟수
    hue='rep',       # 각 반복 실험
    col='tr',        # 전염성 수치
    fit_reg=False,
    palette='viridis'
)

plt.show()
```

```python
cumulative_count = (
    intv_df
    .groupby(['rep', 'intervened', 'tr'])['ig_type']
    .count()
    .groupby(level=['rep']) # 각 반복실험, level을 column형으로 줌
    .cumsum()
    .reset_index()
)

fig = sns.lmplot(
    data=cumulative_count,
    x='intervened',
    y='ig_type',
    hue='rep',
    col='tr',
    fit_reg=False,
    palette='viridis',
    facet_kws={'sharex': False, 'sharey': False, 'legend_out': False}
)

plt.show()
```
